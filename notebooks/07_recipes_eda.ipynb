### EDA - Recipes

```python
import pandas as pd
```

##### Affichage

```python
data = pd.read_csv('../data/raw/recipes.csv')

data.head()
```

##### Dimensions

```python
lignes, columns = data.shape

print(lignes, "lignes")
print(columns, "columns")
```

##### Colonnes

```python
data.columns
```

##### Les column pertinant

```python
data = data[["Name", "TotalTime", "RecipeCategory", "RecipeIngredientParts", "Calories", "FatContent", "SaturatedFatContent", "CholesterolContent", "SodiumContent", "CarbohydrateContent", "FiberContent", "SugarContent", "ProteinContent"]]
```

##### Missing Values

```python
data.isna().sum()
```

##### Remove Recipes with missing values

```python
data = data[data["RecipeCategory"].notnull()]
```

##### Duplicated values

```python
data['Name'].duplicated().sum()
```

##### Remove Duplicated values

```python
data = data.drop_duplicates(subset=["Name"])
```

##### Rows left

```python
data.count()
```

##### Remove plates that take at least an hour

```python
data = data[data["TotalTime"].str.find("H") < 0]
```

##### Remove plates that are prepared instantly

```python
data = data[data["TotalTime"].str.find("S") < 0]
```

##### Remove plates that takes more than 15 minutes to be prepared

```python
data = data[data["TotalTime"].str.replace("PT", '').str.replace("M", '').astype(int) <= 15]
```

##### Remove recipes that contain vodka as ingredient

```python
data = data[data["RecipeIngredientParts"].str.find("vodka") < 0]
```

##### Update column Total Time

```python
data["TotalTime"] = data["TotalTime"].str.replace("PT", '').str.replace("M", '').astype(int)
```

##### Show all categories

```python
print("Total:", len(data['RecipeCategory'].unique()), '\n')

for c in data['RecipeCategory'].unique():
    print(c)
```

##### Keep only relevant categories

```python
categories = [
    "Beverages",
    "Breakfast",
    "Lunch/Snacks",
    "Salad",
    "Soup/Stew",
    "Pasta",
    "Rice",
    "Seafood",
    "Meat",
    "Dessert"
]

data = data[data['RecipeCategory'].isin(categories)]
```

##### Clean the ingredients columns

```python
data['RecipeIngredientParts'] = data['RecipeIngredientParts']           \
    .str.replace("c(", "")  \
    .str.replace('"', '')   \
    .str.replace(')', '')
```

##### Keep recipes with at least 10 ingredients

```python
df_filtered = data[data['RecipeIngredientParts'].str.split(',').str.len() >= 10]
```

##### Keep only 10 rows from each category

```python
recipes = df_filtered.groupby("RecipeCategory").sample(n=10, random_state=42)
```

##### Add Price column

```python
import numpy as np

recipes['Price'] = np.random.randint(5, 51, size=len(recipes))
```

##### Save the result

```python
recipes.to_csv('../data/processed/menu.csv', header=True, index=False)
```
